```python
# GeoPulse: Hyper-Local Retail Mobility Analytics
# Step 1: Import required libraries

import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Set random seed for consistent results
np.random.seed(42)
random.seed(42)

print("GeoPulse environment is ready!")
```

```python
# Step 2: Define 10 retail locations in Visakhapatnam

locations = [
    ["L001", "RK Beach Retail Hub", "RK Beach", 17.7126, 83.3220, "Shopping"],
    ["L002", "Dwaraka Retail Point", "Dwaraka Nagar", 17.7280, 83.3010, "Fashion"],
    ["L003", "MVP Shopping Hub", "MVP Colony", 17.7390, 83.3370, "Grocery"],
    ["L004", "Gajuwaka Market Hub", "Gajuwaka", 17.6868, 83.2165, "Electronics"],
    ["L005", "Jagadamba Retail Point", "Jagadamba Junction", 17.7120, 83.3010, "Fashion"],
    ["L006", "Maddilapalem Shopping Hub", "Maddilapalem", 17.7370, 83.3150, "Grocery"],
    ["L007", "Seethammadhara Retail Hub", "Seethammadhara", 17.7395, 83.3070, "Shopping"],
    ["L008", "NAD Market Point", "NAD Junction", 17.7450, 83.2350, "Electronics"],
    ["L009", "Kurmannapalem Retail Hub", "Kurmannapalem", 17.6670, 83.1910, "Grocery"],
    ["L010", "Pendurthi Shopping Point", "Pendurthi", 17.8290, 83.2050, "Fashion"]
]

# Convert the location list into a DataFrame
locations_df = pd.DataFrame(
    locations,
    columns=[
        "location_id",
        "location_name",
        "area",
        "latitude",
        "longitude",
        "store_category"
    ]
)

# Display all 10 locations
print("GeoPulse - Visakhapatnam Retail Locations")
display(locations_df)
```

```python

# Step 3: Generate 30 days of hourly retail activity

records = []

# Generate 30 days starting from January 1, 2025
dates = pd.date_range(start="2025-01-01", periods=30, freq="D")

# Base footfall for each retail location
location_traffic = {
    "L001": 65,
    "L002": 50,
    "L003": 55,
    "L004": 60,
    "L005": 75,
    "L006": 45,
    "L007": 40,
    "L008": 50,
    "L009": 35,
    "L010": 30
}

# Hourly footfall pattern (24 hours)
hourly_pattern = [
    0.02, 0.01, 0.01, 0.01, 0.01, 0.02,
    0.05, 0.10, 0.20, 0.35, 0.45, 0.55,
    0.65, 0.60, 0.50, 0.45, 0.55, 0.75,
    1.00, 1.10, 0.90, 0.65, 0.35, 0.12
]

# Generate records for every location, date and hour
for _, location in locations_df.iterrows():

    for date in dates:

        day_type = "Weekend" if date.dayofweek >= 5 else "Weekday"

        weekend_factor = 1.25 if day_type == "Weekend" else 1.0

        for hour in range(24):

            weather = np.random.choice(
                ["Sunny", "Cloudy", "Rainy"],
                p=[0.55, 0.30, 0.15]
            )

            weather_factor = {
                "Sunny": 1.0,
                "Cloudy": 0.95,
                "Rainy": 0.75
            }[weather]

            expected_footfall = (
                location_traffic[location["location_id"]]
                * hourly_pattern[hour]
                * weekend_factor
                * weather_factor
            )

            footfall = int(
                np.random.poisson(max(expected_footfall, 0.1))
            )

            # Estimate how many visitors make a purchase
            conversion_rate = 0.30

            transactions = int(
                np.random.binomial(footfall, conversion_rate)
            )

            # Average spending depends on store category
            average_spend = {
                "Shopping": 700,
                "Fashion": 1200,
                "Grocery": 350,
                "Electronics": 2500
            }[location["store_category"]]

            revenue = round(
                transactions * np.random.uniform(
                    average_spend * 0.8,
                    average_spend * 1.2
                ),
                2
            )

            records.append([
                date.date(),
                f"{hour:02d}:00",
                location["location_id"],
                location["location_name"],
                location["area"],
                location["latitude"],
                location["longitude"],
                footfall,
                location["store_category"],
                transactions,
                revenue,
                weather,
                day_type
            ])

# Convert generated records into a DataFrame
df = pd.DataFrame(
    records,
    columns=[
        "date",
        "time",
        "location_id",
        "location_name",
        "area",
        "latitude",
        "longitude",
        "footfall",
        "store_category",
        "transactions",
        "revenue",
        "weather",
        "day_type"
    ]
)

print("Dataset generated successfully!")
print("Total records:", len(df))
display(df.head(10))
```

```python
# Step 4: Basic dataset verification

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nUnique Locations:", df["location_id"].nunique())

print("\nDate Range:")
print(df["date"].min(), "to", df["date"].max())

print("\nTotal Footfall:", df["footfall"].sum())

print("\nTotal Transactions:", df["transactions"].sum())

print("\nTotal Revenue:", round(df["revenue"].sum(), 2))
```

```python
# Step 5: Export the raw dataset as CSV

df.to_csv("geopulse_raw_data.csv", index=False)

print("CSV file created successfully!")
print("File name: geopulse_raw_data.csv")
```
